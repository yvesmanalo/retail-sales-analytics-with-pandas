```python
# Analysis

# Find all customers whose membership is "Gold".

df_customers[df_customers['membership'] == 'Gold']
```

```python
# Find products whose price is greater than 500.

df_products[df_products['price'] > 500]
```

```python
# Find all orders made after 2025-06-01:

df_orders['order_date']= pd.to_datetime(df_orders['order_date'], format = '%Y-%m-%d')
df_orders[df_orders['order_date'] > '2025-06-01']
```

```python
# Sort products by price descending.

df_products.sort_values(by = 'price', ascending = False)
```

```python
# Sort customers alphabetically by name.

df_customers.sort_values(by = 'customer_name')
```

```python
# Set customer_id as index

df_customers.set_index('customer_id', inplace = False)
```

```python
# Retrieve customer with ID 500.

df_customers[df_customers['customer_id'] == 500]
```

```python
# Count customers per membership level.

df_membership_lvl_count = df_customers.groupby('membership').agg({'customer_id':['count']})
df_membership_lvl_count
```

```python
# Find average product price by category.

df_avg_price_per_category = df_products.groupby('category').agg({'price':['mean']})
df_avg_price_per_category
```

```python
# Find number of products per category.

df_num_products_per_category = df_products.groupby('category').agg({'product_id':['count']})
df_num_products_per_category
```

```python
# Merge customers and orders

df_customers.merge(df_orders, on = 'customer_id')
```

```python
# Merge order_items and products

df_order_items.merge(df_products, on = 'product_id')
```

```python
# Create a complete sales table by merging:
# customers
# orders
# order_items
# products

df_merge_customers_orders = df_customers.merge(df_orders, on = 'customer_id')
df_merge_order_items_products = df_order_items.merge(df_products, on = 'product_id')
```

```python
df_master_dataset = df_merge_customers_orders.merge(df_merge_order_items_products, on = 'order_id')
```

```python
# Using the master dataset:
# Calculate total sales per category. (Formula: price × quantity)

df_master_dataset['total_sales'] = df_master_dataset['price'] * df_master_dataset['quantity']
df_master_dataset.groupby('category').agg({'total_sales':'sum'})
```

```python
# Find top 10 selling products.

df_top_10_selling_products = df_master_dataset.groupby('product_name').agg({'total_sales':'sum'}).sort_values(by = 'total_sales', ascending = False).iloc[:10]
df_top_10_selling_products
```

```python
# Find top 10 customers by spending.

df_top_10_spenders = df_master_dataset.groupby('customer_name').agg({'total_sales':'sum'}).sort_values(by = 'total_sales', ascending = False).iloc[:10]
df_top_10_spenders
```

```python
# Find sales by city.

df_sales_per_city = df_master_dataset.groupby('city').agg({'total_sales':'sum'})
df_sales_per_city
```

```python
# Convert 'customer_id' to index in 'customers' and 'orders' dataframes.
# Use join() instead of merge().

df_join_customers_orders = df_customers.set_index('customer_id').join(df_orders.set_index('customer_id'))
df_join_customers_orders
```

```python
# Concatenate 'customers_part1.csv' and 'customers_part2.csv'
pd.concat([df_customers_part1, df_customers_part2], axis = 0)
```

```python
# Split products by category and concatenate them back together.

# Products by 'Sports' category
df_sports_category = df_products[df_products['category'] == 'Sports']
df_sports_category
```

```python
# Concatenation of product by categories dataframes

df_concat_product_categories = pd.concat([df_books_category, df_clothing_category, df_electronics_category, df_sports_category, df_home_category], axis = 0)
df_concat_product_categories.sort_values(by = 'product_id', ascending = True)
```

```python
# Further Analysis

# Which category generated the highest revenue?

df_master_dataset.groupby('category').agg({'total_sales':'sum'}).sort_values(by = 'total_sales', ascending = False).iloc[:1]
```

```python
# Which customer spent the most money?

df_top_10_spenders.iloc[:1]
```

```python
# What is the average order value? (Formula: Total Revenue / Number of Orders)

#Considering that the sales is only the source of revenue.

avg_order_value = df_master_dataset['total_sales'].sum()/df_master_dataset['order_id'].count()
avg_order_value
```

```python
# Which city generated the highest revenue?

df_master_dataset.groupby('city').agg({'total_sales':'sum'}).iloc[:1]
```

```python
# What are the monthly sales trends?

month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

df_master_dataset['order_month'] = df_master_dataset['order_date'].dt.month_name()
df_master_dataset['order_month'] = pd.Categorical(df_master_dataset['order_month'], categories = month_order, ordered = True)
df_master_dataset.groupby('order_month', observed = True).agg({'total_sales':'sum'})




```
